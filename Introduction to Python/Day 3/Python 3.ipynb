```python
def sum_dig(n):
    """
    Function to calculate the sum of the digits of a number.

    Args:
        n (int): The number whose digits will be summed.

    Returns:
        int: The sum of the digits of the number.
    """
    n = int(input("Enter the integer for sum of digits: "))
    return sum(int(digit) for digit in str(abs(n)))


sum_dig(999)
```

```python
def closest(lst, n):
    for i in lst:
        mx = max(lst)
        if mx <= n:
            return mx
        else:
            lst.remove(mx)
            continue


closest([1, 6, 3, 9, 11], 8)
```
