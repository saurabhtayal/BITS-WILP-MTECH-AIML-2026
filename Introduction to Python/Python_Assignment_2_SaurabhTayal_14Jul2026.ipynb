4. Write a Python program to create a 1x2 grid of subplots. The first subplot should show a simple  line plot with data [1, 2, 3, 4] and [10, 20, 30, 40]. The second subplot should display a bar plot  with the same x-axis values and corresponding y-axis values [5, 10, 15, 20]. 

```python
fig, axes = plt.subplots(1, 2, figsize=(10, 4))

axes[0].plot([1, 2, 3, 4], [10, 20, 30, 40], marker="o")
axes[0].set_title("Line Plot")
axes[0].set_xlabel("X")
axes[0].set_ylabel("Y")

axes[1].bar([1, 2, 3, 4], [5, 10, 15, 20], color="purple")
axes[1].set_title("Bar Plot")
axes[1].set_xlabel("X")
axes[1].set_ylabel("Y")

plt.tight_layout()
plt.show()
```

6. Create a Python program that defines a custom function to calculate the factorial of a number.  The function should take one integer as an argument and return its factorial. 
Sample Input -> 5 
Sample Output -> The factorial of 5 is: 120 


```python
def factorial(n):
    result = 1
    for i in range(1, n + 1):
        result *= i
    return result


num = 5
print("The factorial of", num, "is:", factorial(num))
```

9. you are working as a data analyst in a college where the academic coordinator wants to evaluate how students perform across different subjects and sections. You are given a dataset containing Student, Subject, Marks, and Section. Using Seaborn, you need to create a catplot() to visualize the average marks for each subject, while using hue="Section" to compare performance between different sections. You should experiment with both "bar" and "point" plot types to understand how each represents the data differently. Enhance the visualization by adding a meaningful title such as “Subject-wise Performance by Section”, choosing an appropriate color palette, and increasing the plot size for clarity. After generating the plots, analyze them to determine which subject shows the highest variation in marks across sections and interpret what this suggests about student performance.

```python
# Sample dataset
student_data = pd.DataFrame(
    {
        "Student": ["A", "B", "C", "D", "E", "F", "G", "H", "I", "J", "K", "L"],
        "Subject": [
            "Math",
            "Math",
            "Math",
            "Science",
            "Science",
            "Science",
            "English",
            "English",
            "English",
            "History",
            "History",
            "History",
        ],
        "Marks": [85, 78, 90, 88, 82, 91, 76, 80, 84, 89, 87, 92],
        "Section": ["A", "B", "A", "B", "A", "B", "A", "B", "A", "B", "A", "B"],
    }
)

student_data
```

```python
# Average marks by subject and section
avg_marks = student_data.groupby(["Subject", "Section"])["Marks"].mean().reset_index()
avg_marks
```

```python
plt.figure(figsize=(12, 6))
sns.catplot(
    data=avg_marks,
    x="Subject",
    y="Marks",
    hue="Section",
    kind="bar",
    palette="Set2",
    height=5,
    aspect=1.6,
)
plt.gcf().suptitle("Subject-wise Performance by Section", y=1.02)
plt.show()

plt.figure(figsize=(12, 6))
sns.catplot(
    data=avg_marks,
    x="Subject",
    y="Marks",
    hue="Section",
    kind="point",
    palette="Set2",
    height=5,
    aspect=1.6,
)
plt.gcf().suptitle("Subject-wise Performance by Section", y=1.02)
plt.show()

# Interpretation
subject_variation = (
    avg_marks.groupby("Subject")["Marks"].var().sort_values(ascending=False)
)
print("Variation across sections by subject:")
print(subject_variation)
print("\nHighest variation subject:", subject_variation.idxmax())
```

10. You are analyzing sales data for a retail company operating across multiple regions. The dataset includes Product, Region, and Sales. Your task is to use catplot(kind="bar") in Seaborn to compare product sales across regions, using hue="Product" to differentiate between products. Additionally, apply col="Region" to create separate visualizations for each region, enabling clearer comparisons. Customize the plot using the "Set2" color palette and adjust the height and aspect to improve readability. Once the visualization is complete, analyze the patterns to identify which product performs consistently well across all regions and explain the possible reasons behind this trend.

```python
sales_data = pd.DataFrame(
    {
        "Product": [
            "Laptop",
            "Phone",
            "Tablet",
            "Laptop",
            "Phone",
            "Tablet",
            "Laptop",
            "Phone",
            "Tablet",
        ],
        "Region": [
            "North",
            "North",
            "North",
            "South",
            "South",
            "South",
            "East",
            "East",
            "East",
        ],
        "Sales": [1200, 800, 500, 1300, 750, 550, 1100, 850, 600],
    }
)

plt.figure(figsize=(14, 5))
sns.catplot(
    data=sales_data,
    x="Region",
    y="Sales",
    hue="Product",
    kind="bar",
    col="Region",
    palette="Set2",
    height=4,
    aspect=1.1,
)
plt.suptitle("Sales by Product Across Regions", y=1.03)
plt.show()

product_summary = (
    sales_data.groupby("Product")["Sales"].mean().sort_values(ascending=False)
)
print("Average sales by product:")
print(product_summary)
print("\nConsistently strong product:", product_summary.idxmax())
```
