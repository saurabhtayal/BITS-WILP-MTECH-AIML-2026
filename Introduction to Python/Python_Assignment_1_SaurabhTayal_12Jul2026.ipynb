abcd... = an + bn + cn + dn + ...

153 is an Armstrong number of order 3.

3. Given a string, write a python function to check if it is  palindrome or not. A string is said to be palindrome if the reverse of the string is the same as string. For example, “malayalam” is a palindrome, but “music” is not a palindrome.

```python
def palin(str):
    if str == str[::-1]:
        print("Palindrome")
    else:
        print("Not a Palindrome")


palin("malayalam")
```

```python
palin("music")
```

4. Given an array which may contain duplicates, print all elements and their frequencies.

```python
arr = [1, 2, 3, 4, 5, 5, 4, 3, 2, 1]
lmnts = []
cnt = []
for i in range(len(arr)):
    if arr[i] not in lmnts:
        lmnts.append(arr[i])
        cnt.append(1)
    else:
        idx = lmnts.index(arr[i])
        cnt[idx] += 1
print("Elements: ", lmnts)
print("Count: ", cnt)
```

5. Given a number n, write a function to print all prime factors of n. For example, if the input number is 12, then output should be “2 2 3”.

```python
def prime_fac(n):
    chk = n
    prime = []
    while chk > 1:
        for i in range(2, n):
            if chk % i == 0:
                prime.append(i)
                chk = chk / i
                break
            else:
                continue
    print("Prime factors of ", n, " are: ", prime)


prime_fac(12)
```

```python
prime_fac(24)
```

6. Given two numbers n and r, find the value of nCr (binomial coefficient: nCr = (n!) / (r! * (n-r)!))

```python
def bi_coef(n, r):
    if r < n:
        num = 1
        den1 = 1
        den2 = 1
        for i in range(n):
            num = num * (n - i)
        for i in range(r):
            den1 = den1 * (r - i)
        for i in range(n - r):
            den2 = den2 * (n - r - i)
        print("Numerator is: ", num)
        print("Denominator 1 is: ", den1)
        print("Denominator 2 is: ", den2)
        print("The binomial coefficient is: ", num / (den1 * den2))
    else:
        print("Not possible")


bi_coef(5, 2)
```

7. Searching: Given a sorted array arr[] of n elements, write a function to search a given element x in arr[]. Do it using linear and binary search techniques.

```python
# Gives the target elements' all actual index positions in any unsorted list using linear search algorithm as its base concept
def linear_search(lst, target):
    arr = []
    for i in range(len(lst)):
        if lst[i] == target:
            arr.append(i)
    if len(arr) > 0:
        print("Element found at index(s): ", arr)
    else:
        print("Element not found in the list.")


numbers = [10, 50, 30, 30, 70, 80, 20]
target_value = 30

linear_search(numbers, target_value)
```

9. A hospital is planning to digitize its patient record system to improve efficiency in managing patient information and billing. As a Python developer, you are tasked with building a simple system to process and analyze patient data. You are given a collection of patient details, where each record contains the patient’s name, age, disease, and total bill amount.
•	Store and organize patient data using appropriate data structures 
•	Calculate the number of patients affected by each disease 
•	Identify the patient with the highest bill amount 
•	Compute the average bill and display patients whose bill exceeds the average 
•	Ensure that duplicate patient entries are removed using suitable techniques 

```python
import pandas as pd

patients = [
    {"name": "Asha", "age": 45, "disease": "Fever", "bill": 1200},
    {"name": "Ravi", "age": 33, "disease": "Flu", "bill": 1800},
    {"name": "Meera", "age": 28, "disease": "Diabetes", "bill": 2500},
    {"name": "Asha", "age": 45, "disease": "Fever", "bill": 1200},
    {"name": "Karan", "age": 50, "disease": "Flu", "bill": 1400},
    {"name": "Neha", "age": 36, "disease": "Asthma", "bill": 2100},
]

patient_df = pd.DataFrame(patients)
patient_df
```

```python
patient_df = patient_df.drop_duplicates().reset_index(drop=True)
patient_df
```

```python
# Count patients affected by each disease
disease_count = patient_df["disease"].value_counts()
disease_count
```

```python
patient_df["bill"].idxmax()
```

```python
# Find the patient with the highest bill
highest_bill_patient = patient_df.loc[patient_df["bill"].idxmax()]
highest_bill_patient
```

```python
# Compute average bill and list patients above average
average_bill = patient_df["bill"].mean()
average_bill
```

```python
above_average = patient_df[patient_df["bill"] > average_bill]
above_average
```

```python
print("Unique patients:\n", patient_df)
print("\nDisease count:\n", disease_count)
print("\nPatient with highest bill:\n", highest_bill_patient)
print("\nAverage bill:", average_bill)
print("\nPatients above average:\n", above_average)
```

2. Online Food Delivery Analysis System
An online food delivery platform wants to analyze customer orders to understand business performance and customer preferences. Each order includes item names, quantities, and the restaurant from which the order was placed. As a developer, your task is to build a program that processes this data and generates meaningful insights.
•	Store order details in a structured format 
•	Calculate total revenue generated by each restaurant 
•	Identify the most frequently ordered item 
•	Detect and remove duplicate orders, if present 
•	Display the top-performing restaurant based on revenue 

```python
import pandas as pd

orders = [
    {"item": "Pizza", "quantity": 2, "restaurant": "Domino's"},
    {"item": "Burger", "quantity": 1, "restaurant": "Burger King"},
    {"item": "Pizza", "quantity": 2, "restaurant": "Domino's"},
    {"item": "Fries", "quantity": 3, "restaurant": "Burger King"},
    {"item": "Pasta", "quantity": 1, "restaurant": "Pizza Hut"},
    {"item": "Burger", "quantity": 1, "restaurant": "Burger King"},
]

order_df = pd.DataFrame(orders)
order_df
```

```python
order_df = order_df.drop_duplicates().reset_index(drop=True)
order_df
```

```python
# Calculate revenue by restaurant
prices = {"Pizza": 250, "Burger": 180, "Fries": 80, "Pasta": 200}
order_df["revenue"] = order_df["item"].map(prices) * order_df["quantity"]
order_df
```

```python
restaurant_revenue = order_df.groupby("restaurant")["revenue"].sum()
restaurant_revenue
```

```python
# Most frequently ordered item
item_count = order_df["item"].value_counts()
item_count
```

```python
item_count.idxmax()
```

```python
item_count.max()
```

```python
most_frequent_item = item_count.idxmax(), item_count.max()
most_frequent_item
```

```python
# Top-performing restaurant
top_restaurant = restaurant_revenue.idxmax(), restaurant_revenue.max()
top_restaurant
```

```python
print("Unique orders:\n", order_df)
print("\nRevenue by restaurant:\n", restaurant_revenue)
print("\nMost frequently ordered item:", most_frequent_item)
print("\nTop-performing restaurant:", top_restaurant)
```
